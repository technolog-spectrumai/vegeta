cruise: **T-Motor FA24.2 folding** on MN801-S. 12S6P pack, the middle wing-root hardware (`root_M`).

## Reading it

- **Payload and range**: Hanna 3 carries 3.5 kg 84 km (R3 radius 20.5 km), Joanna 3 4.5 kg 81 km (19.0 km) — Ginna 3's
  83 km kept. The price is mass: 21.4 and 22.7 kg (Ginna 3: 12.5 kg): the 12S6P pack (5.4 kg), the bigger airframe
  (×1.74 / ×1.77 of Freddy, to keep the 18 m/s stall), the middle root hardware. Both stay under the 24 kg line.
- **Hover**: T/W 1.70 / 1.60, 14 / 13 min to an empty pack with the payload (Ginna 3: 11 min), 4.0 / 4.3 kW.
- **Speed traded**: 115 km/h top (Ginna 3: 144), 0–100 km/h in 5.3 / 5.6 s (2.8).
- **Open**: the CAD at these scales; the stall at 18.1 m/s on Joanna (0.1 over the cap); the rigid front props'
  ground and transition clearances; above 20 kg the operation category (an authorisation in the specific category).
