## 3. The CAD of K4TQ-F (the sections lofted on each wing)

```python
import k4tq_cad
vF = evals["F"]["veh"]
png = k4tq_cad.render(kc.DATA / "figures" / "k4tq_F.png", vF["p"], info={"mass": vF["mass"]["mass_kg"]}, items=vF["items"])
Image(str(png))
```

**Reading it.** The sections matter less in the cruise than the polars suggest (the front wing's high cl sits where the
cambered 15 % and the 12 % sections are close) but they set the stall: F's NACA 4415 front wing reaches a higher cl_max
at the transition's Reynolds number, which lowers the aircraft's stall speed and lets it cruise slower. The hard
constraint is the spar: NACA 2412 on the 120 mm front tip chord leaves 14 mm of depth for a Ø16 spar that needs 18.
