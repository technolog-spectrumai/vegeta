## Reading it — does K4TQ beat a quadcopter of the same size?

**Yes on the mission it was built for, no in the hover — and by less than the same-pack numbers suggest.**

- **Same pack (6S3P)**: Cathy and Freddy reach ≈14 km R3 radius with 1 kg; Q-13 and Q-17 ≈6.5 km — the quads do
  not close the 10 km requirement. Still-air range ≈60 km against ≈23 km: the wing flies at L/D ≈10 and needs
  ≈330 W at 23 m/s where a quad needs ≈580–690 W at its best-range speed.
- **Same take-off mass (the fair one)**: the tilt-wing hardware (roots, wings, spars, nacelles, a longer fuselage)
  weighs ≈2.4 kg. Spent on cells, Q-17 carries 6S7P and reaches ≈12.7 km — it closes R3 too, ≈8 % short of K4TQ.
  The price: 42 cells against 18 (cost, charging, dangerous-goods shipping), T/W only 1.65 (it climbs and turns worst
  of all), and every extra km costs it twice K4TQ's energy, so in the 8 m/s wind, or with any reserve loss, the gap
  grows. On the 0.5 kg extended mission K4TQ keeps ≈15 km against ≈13.8.
- **Hover**: the quads win clearly (Q-17 25 min on the same pack, 36 min heavy, K4TQ 13 min): bigger, flatter discs,
  no wing in the downwash. A mission with long hovers belongs to a quad.
- **Agility**: the light quads roll and pitch about twice as hard in the hover and turn tighter at low speed; K4TQ
  yaws 2–3× harder in the hover (wings tilted against each other beats rotor reaction torque) and turns at 40–44 deg/s
  at 20–30 m/s on its wings.
- **Speed**: K4TQ 28 m/s on its drive; the quads reach the model's sweep end (30 m/s) — optimistic, the edgewise
  propeller losses at 30° disc tilt are not modelled.

**Caveats.** The quad structure (pod, arms) is a light first estimate and gets no CAD; the BEMT maps are coarsened under
the mocks; the quads' forward-flight model (Glauert + μ² profile growth) has not been checked against `08_quadcopter`'s
flight data. Freddy and Cathy are the same here except in the climb and the turn (Freddy's 4415 front wing).
