## Reading it

- **K4TQ wins the distance**: the R3 radius 12–18 km against 4–6 km, still-air range 54–82 km against 16–20 km, and a
  higher mission-average speed (62–67 km/h against 44–48 km/h: the FPV quad must fly slowly to fly far).
- **The FPV quad wins the hover** (17–24 min against 10–14) and the sprint off the hover (1.9–2.7 s to 100 km/h with
  payload against 2.7–2.9 s for F3 / G3; F2 needs 5.2 s — its T/W is 1.6). K4TQ's sprint is limited by the wing's tilt
  (90° in 2 s) and by the hover thrust that must carry the weight until the wing does.
- **Top speed**: F3 162 km/h and G3 144 km/h beat the FPV's 121–124 km/h with payload (the 3 s figure equals the 180 s
  one: the drives run at their 180 s peak current; there is no thermal model).
- **Straight up**: nobody reaches 100 km/h — K4TQ tops out at 42–62 km/h (the four discs carry the weight edge-on wings
  and a broadside fuselage), the FPV quad at 72–96 km/h.
